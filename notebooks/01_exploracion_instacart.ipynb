### Filtros con condiciones simultáneas

```python
# Pedidos a las 10 horas Y con order_number > 10
f1 = orders.loc[(orders["order_hour_of_day"] == 10) & (orders["order_number"] > 10)]
display(f1.head()); print(len(f1))
```

Con & se exigen ambas condiciones: quedan 21.537 pedidos, menos que con cada condición por separado.

```python
# Productos del departamento 1 Y con aisle_id > 50
f2 = products.loc[(products["department_id"] == 1) & (products["aisle_id"] > 50)]
display(f2.head()); print(len(f2))
```

Con & se exigen ambas condiciones: quedan 1.618 productos, menos que los 4.008 del departamento 1.


### Preguntas de análisis (Momento 4)
- **¿Selección de columnas vs. filtrado de filas?** R=Seleccionar columnas escoge qué atributos ver (ejercicio 1) y reduce el ancho de la tabla. Filtrar filas escoge qué registros cumplen una condición y reduce el alto, conservando todas las columnas.
- **¿Ventajas de loc[]?** R=Permite indicar en una sola instrucción qué filas (con una condición) y qué columnas se quieren. Es explícito y legible.
- **¿Qué pasa con la cantidad de registros al aplicar varias condiciones?** _(compara f1 con ej3 y ej5) R=Cada condición adicional mantiene o reduce el resultado. Pasé de 40.578 pedidos a las 10 h (ej3) a 21.537 al exigir también order_number > 10 (f1), y de 4.008 productos del departamento 1 (ej2) a 1.618 al exigir aisle_id > 50 (f2).
- **¿Por qué usar bien los operadores lógicos?** _(pista: `&`, `|`, paréntesis) R=En Pandas se usa & (y) y | (o), con cada condición entre paréntesis. Con & se obtienen las filas que cumplen ambas condiciones. Con | se obtienen las que cumplen al menos una, que da un resultado muy distinto.
